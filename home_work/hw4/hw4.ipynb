# Путешествие в страну Х
## Задача:

Вы отправляетесь в страну X, которая славится своими мистическими приключениями. Вам предстоит пройти через две опасные локации: лес и горы.

В лесу есть две возможные тропы - тропа A и тропа B. Вы можете выбрать одну из них случайным образом. Вероятность выбрать тропу A равна 0.4, а тропу B - 0.6. После выбора тропы вы перемещаетесь в горы. Вероятность успешно пройти горы, если вы выбрали тропу A, составляет 0.3. Если же вы выбрали тропу B, вероятность успешного прохождения гор увеличивается до 0.7.


## Вопрос:

Какова вероятность успешно пройти оба этапа путешествия (попасть в страну X) и как она зависит от выбора тропы в лесу?



```python
p_a = 0.4 # Вероятность выбрать тропу A
p_b = 0.6 # Вероятность выбрать тропу B
p_a_success = 0.3 # Вероятность успешно пройти горы по тропе A
p_b_success = 0.7 # Вероятность успешно пройти горы по тропе B
```

```python
p_success = p_a * p_a_success + p_b * p_b_success
print(f"Вероятность успешно пройти горы: {p_success:.2f}")
```

# Лабиринт сокровищ
## Задача:

Вы оказались в лабиринте, который славится своими сокровищами. Чтобы добраться до сокровищ, вам нужно пройти через три замка: замок A, замок B и замок C. Каждый замок защищен различными ловушками и подходы к каждому из них различны.
- 
- Замок A: Первый замок защищен огненными ловушками. Вероятность успешно пройти замок A составляет 0.6.
- Замок B: После успешного прохождения замка A, вы переходите к замку B. Замок B защищен сложными головоломками. Если вы успешно прошли замок A, вероятность успешно пройти замок B составляет 0.4. Однако, если вы не смогли пройти замок A, то вероятность успешно пройти замок B уменьшается до 0.2.
- Замок C: После прохождения замка B, вам предстоит столкнуться с последней преградой - замком C. Замок C охраняется могучим воином. Если вы успешно прошли замок B, вероятность успешно пройти замок C составляет 0.7. Однако, если вы не смогли пройти замок B, то вероятность успешно пройти замок C уменьшается до 0.3.

## Вопрос:

Какова вероятность добраться до сокровища (пройти все три замка) и как она зависит от вашего успешного прохождения предыдущих замков?

```python
p_A_s = 0.6 # Вероятность пройти замок А
p_B_sA = 0.4 # Вероятность пройти замок B, если прошел замок A
p_B_nA = 0.2 # Вероятность пройти замок B, если не прошел замок A
p_C_sB = 0.7 # Вероятность пройти замок C, если прошел замок B
p_C_nB = 0.3 # Вероятность пройти замок C, если не прошел замок B
```

```python
# Если подойти формально к формулировке вопроса, то это будет совместная веротность трех событий
# Пройти замок A, пройти замок B, пройти замок C
p_ABC = p_A_s * p_B_sA * p_C_sB
print(f"Вероятность пройти все три замка: {p_ABC:.2f}") 
```

```python
# Но столь длинная формулировка намекает, что успех это пройти именно последний замок, и не важно прошел ли ты первые 2
p_B_s = p_A_s * p_B_sA + (1 - p_A_s) * p_B_nA #  Вероятность пройти замок B
p_B_n = 1 - p_B_s # Вероятность не пройти замок B
p_C_s = p_B_s * p_C_sB + p_B_n * p_C_nB # Вероятность пройти замок C
print(f"Вероятность пройти замок C: {p_C_s:.2f}")
```

В чем я вижу разницу между первым и вторым вариантом?
- «Пройти все три замка» — это событие $A \cap B \cap C$, его вероятность $P(A)P(B\mid A)P(C\mid B)$, мы получили 0.17.
- «Пройти последний замок C» — это событие $C$, и к нему ведут несколько сценариев (через успешный или неуспешный B и А), поэтому используем закон полной вероятности по $B,\neg B$, получили 0.43.

# Заболевание и результаты тестирования
## Задача:

Представим, что у определенной популяции есть редкое заболевание, которое встречается у 1% людей. Существует медицинский тест для определения этого заболевания. Тест имеет высокую точность: вероятность правильно определить наличие заболевания у больного составляет 0.95, а вероятность правильно определить отсутствие заболевания у здорового составляет 0.90.


# Вопрос:

Если у человека положительный тест на заболевание, какова вероятность того, что он действительно болен этим заболеванием?

```python
p_P = 0.01 # Вероятность заболевания
p_Tp_P = 0.95 # Вероятность положительного теста при заболевании
p_Tn_N = 0.90 # Вероятность отрицательного теста при отсутствии заболевания
p_Tn_P = 1 - p_Tp_P # Вероятность отрицательного теста при заболевании
```

```python
# нам нужно найти p_P_Tp - вероятность заболевания при положительном тесте
# p_P_Tp = p_P * p_Tp_P / p_Tp, где p_Tp - вероятность положительного теста
p_Tp = p_P * p_Tp_P + (1 - p_P) * p_Tn_P
p_P_Tp = p_P * p_Tp_P / p_Tp
print(f"Вероятность заболевания при положительном тесте: {p_P_Tp:.2f}")
```

# Студенты и предметы
## Задача:

В университете есть два курса по математике: Алгебра и Анализ. Студенты изучают оба предмета, и 70% студентов посещают курс Алгебры, а 30% студентов посещают курс Анализа. Вероятность того, что студент с курса Алгебры сдаст экзамен, составляет 80%, а вероятность того, что студент с курса Анализа сдаст экзамен, составляет 70%.


## Вопрос:

Если студент сдал экзамен по математике. Какова вероятность того, что он изучал курс Алгебры?



```python
p_Al = .7 # Вероятность того, что студент посещает Алгебру
p_An = .3 # Вероятность того, что студент не посещает Анализ
p_s_Al = .8 # Вероятность того, что студент сдал экзамен, если он посещает Алгебру
p_s_An = .7 # Вероятность того, что студент сдал экзамен, если он посещает Анализ
```

```python
# Нам надо найти p_Al_s - вероятность того, что студент посещает Алгебру при условии, что он сдал экзамен
# p_Al_s = p_Al * p_s_Al / p_s, где p_s - вероятность того, что студент сдал экзамен
p_s = p_Al * p_s_Al + p_An * p_s_An
p_Al_s = p_Al * p_s_Al / p_s
print(f"Вероятность того, что студент посещает Алгебру при условии, что он сдал экзамен: {p_Al_s:.2f}")
```
